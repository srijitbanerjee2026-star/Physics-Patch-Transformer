```python
import os
import pandas as pd

print("Step 1: Setting up workspace directories...")
os.makedirs("data", exist_ok=True)
os.makedirs("models", exist_ok=True)
os.makedirs("utils", exist_ok=True)

# Official URLs for ETTm1 (Dataset A - Train) and ETTm2 (Dataset B - Test)
train_url = "https://raw.githubusercontent.com/zhouhaoyi/ETDataset/main/ETT-small/ETTm1.csv"
test_url = "https://raw.githubusercontent.com/zhouhaoyi/ETDataset/main/ETT-small/ETTm2.csv"

print("Downloading Dataset A (ETTm1) and Dataset B (ETTm2)...")
train_df = pd.read_csv(train_url)
test_df = pd.read_csv(test_url)

# Save them locally inside your Colab session's data folder
train_df.to_csv("data/ETTm1.csv", index=False)
test_df.to_csv("data/ETTm2.csv", index=False)

print("\n--- Setup Complete ---")
print(f"Dataset A Shape (Training): {train_df.shape}")
print(f"Dataset B Shape (Testing): {test_df.shape}")
print("\nFirst 3 rows of Dataset A:")
display(train_df.head(3))
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler

print("Step 2: Normalizing and creating sliding windows (corrected index)...")

# 1. Drop the 'date' string column
train_data = train_df.drop(columns=['date']).values
test_data = test_df.drop(columns=['date']).values

# 2. Fit scaler ONLY on training data to prevent data leakage
scaler = StandardScaler()
train_scaled = scaler.fit_transform(train_data)
test_scaled = scaler.transform(test_data)

print("Data successfully normalized!")

# 3. Function to create sliding window tensors for time-series forecasting
def create_windows(data, input_window=96, pred_window=48):
    X, Y = [], []
    for i in range(len(data) - input_window - pred_window + 1):
        # Input sequence (historical telemetry across all 7 features)
        X.append(data[i : i + input_window])
        # Target sequence: Oil temperature is at index 6 (the last column)
        Y.append(data[i + input_window : i + input_window + pred_window, 6])
    return np.array(X), np.array(Y)

# Define window sizes (96 steps input = 24 hours, 48 steps predict = 12 hours)
INPUT_WINDOW = 96
PRED_WINDOW = 48

X_train, Y_train = create_windows(train_scaled, INPUT_WINDOW, PRED_WINDOW)
X_test, Y_test = create_windows(test_scaled, INPUT_WINDOW, PRED_WINDOW)

print(f"X_train Shape (Batch, Time, Features): {X_train.shape}")
print(f"Y_train Shape (Batch, Prediction_Horizon): {Y_train.shape}")
print(f"X_test Shape (Batch, Time, Features): {X_test.shape}")
print(f"Y_test Shape (Batch, Prediction_Horizon): {Y_test.shape}")
```

```python
import torch
import torch.nn as nn

print("Step 3: Defining the Patch Embedding module...")

class PatchEmbedding(nn.Module):
    def __init__(self, patch_size=16, in_channels=7, d_model=128):
        super().__init__()
        self.patch_size = patch_size
        # Each patch flattens patch_size consecutive steps across all in_channels
        self.projection = nn.Linear(patch_size * in_channels, d_model)

    def forward(self, x):
        # x shape: [Batch, Time_Steps, Channels] -> e.g., [32, 96, 7]
        batch_size, time_steps, channels = x.shape

        # Ensure time steps are divisible by patch size
        num_patches = time_steps // self.patch_size

        # Reshape to group steps into patches: [Batch, Num_Patches, Patch_Size * Channels]
        x = x.reshape(batch_size, num_patches, self.patch_size * channels)

        # Project linear dimension to our model size (d_model)
        x = self.projection(x)
        return x

# Test the patch embedding module with a dummy batch tensor to verify shapes
dummy_input = torch.randn(32, 96, 7) # Batch of 32, 96 time steps, 7 features
patch_layer = PatchEmbedding(patch_size=16, in_channels=7, d_model=128)
dummy_output = patch_layer(dummy_input)

print(f"Dummy Input Shape: {dummy_input.shape}")
print(f"Patched Output Shape (Batch, Num_Patches, d_model): {dummy_output.shape}")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

print("Step 4: Defining the Linear Attention module...")

class LinearAttention(nn.Module):
    def __init__(self, d_model):
        super().__init__()
        self.query = nn.Linear(d_model, d_model)
        self.key = nn.Linear(d_model, d_model)
        self.value = nn.Linear(d_model, d_model)

    def forward(self, x):
        # Apply linear projections for Q, K, V
        Q = self.query(x)
        K = self.key(x)
        V = self.value(x)

        # Apply a positive activation function to act as a kernel approximation
        Q = F.elu(Q) + 1.0
        K = F.elu(K) + 1.0

        # Compute linear attention: changes complexity from O(N^2) to O(N)
        kv_context = torch.matmul(K.transpose(-1, -2), V)
        attention_output = torch.matmul(Q, kv_context)

        return attention_output

# Test the linear attention layer with our previous patch output
attn_layer = LinearAttention(d_model=128)
attn_output = attn_layer(dummy_output)

print(f"Patched Input Shape: {dummy_output.shape}")
print(f"Linear Attention Output Shape: {attn_output.shape}")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

print("Step 5: Building the Full Model and Physics-Informed Loss...")

class PhysicsInformedPatchTransformer(nn.Module):
    def __init__(self, patch_size=16, in_channels=7, d_model=128, num_patches=6, pred_window=48):
        super().__init__()
        # 1. Patch Embedding Layer
        self.patch_embed = PatchEmbedding(patch_size, in_channels, d_model)

        # 2. Linear Attention & Transformer Block components
        self.attention = LinearAttention(d_model)
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)

        self.feed_forward = nn.Sequential(
            nn.Linear(d_model, d_model * 2),
            nn.GELU(),
            nn.Linear(d_model * 2, d_model)
        )

        # 3. Forecasting Head (Maps flattened latent patches to the 48-step prediction horizon)
        self.head = nn.Sequential(
            nn.Flatten(),
            nn.Linear(num_patches * d_model, 256),
            nn.GELU(),
            nn.Linear(256, pred_window)
        )

    def forward(self, x):
        # x shape: [Batch, 96, 7]
        # Step A: Patch embedding -> [Batch, 6, 128]
        x = self.patch_embed(x)

        # Step B: Linear Attention with residual connection and layer norm
        attn_out = self.attention(x)
        x = self.norm1(x + attn_out)

        # Step C: Feed-forward block with residual connection
        ff_out = self.feed_forward(x)
        x = self.norm2(x + ff_out)

        # Step D: Final prediction projection -> [Batch, 48]
        output = self.head(x)
        return output

# --- Physics-Informed Thermodynamic Loss ---
def physics_informed_loss(predictions, targets, inputs):
    """
    Penalizes violations of heat-transfer dynamics.
    Enforces that rapid changes in electrical load features (e.g., index 0, HUFL)
    correlate proportionally with changes in oil temperature predictions.
    """
    # Standard Data-driven MSE Loss
    mse_loss = F.mse_loss(predictions, targets)

    # Thermodynamic Residual: Rate of change of predicted temperature vs. electrical load input
    # inputs shape: [Batch, Time, Features], taking mean load across features as proxy
    load_proxy = inputs[:, :, 0].mean(dim=1) # [Batch]
    pred_temp_slope = predictions[:, -1] - predictions[:, 0] # [Batch] change over horizon

    # Physics constraint: Temperature rise should follow thermal inertia relative to load
    physics_residual = torch.mean(F.relu(torch.abs(pred_temp_slope) - (load_proxy * 2.0)))

    # Combined Loss (Alpha balances data accuracy vs physical law adherence)
    alpha = 0.05
    total_loss = mse_loss + (alpha * physics_residual)
    return total_loss

# Test the full architecture with a dummy batch
model = PhysicsInformedPatchTransformer()
sample_input = torch.randn(32, 96, 7)
sample_target = torch.randn(32, 48)

preds = model(sample_input)
loss = physics_informed_loss(preds, sample_target, sample_input)

print(f"Model Forward Output Shape: {preds.shape}")
print(f"Calculated Physics-Informed Loss Value: {loss.item():.4f}")
```

```python
class RevIN(nn.Module):
    def __init__(self, num_features: int, eps=1e-5, affine=True):
        super().__init__()
        self.num_features = num_features
        self.eps = eps
        self.affine = affine
        if self.affine:
            self.affine_weight = nn.Parameter(torch.ones(num_features))
            self.affine_bias = nn.Parameter(torch.zeros(num_features))

    def forward(self, x, mode: str):
        if mode == 'norm':
            self._get_statistics(x)
            x = self._normalize(x)
        elif mode == 'denorm':
            x = self._denormalize(x)
        else:
            raise ValueError(f"Unknown mode: {mode}")
        return x

    def _get_statistics(self, x):
        dim = 1
        self.mean = x.mean(dim=dim, keepdim=True).detach()
        self.stdev = torch.sqrt(x.var(dim=dim, keepdim=True, unbiased=False) + self.eps).detach()

    def _normalize(self, x):
        x = (x - self.mean) / self.stdev
        if self.affine:
            x = x * self.affine_weight + self.affine_bias
        return x

    def _denormalize(self, x):
        mean_ot = self.mean[:, 0, 6]
        stdev_ot = self.stdev[:, 0, 6]
        if self.affine:
            weight_ot = self.affine_weight[6]
            bias_ot = self.affine_bias[6]
            x = (x - bias_ot) / (weight_ot + 1e-8)
        x = x * stdev_ot.unsqueeze(1) + mean_ot.unsqueeze(1)
        return x

class PhysicsInformedPatchTransformerRevIN(nn.Module):
    def __init__(self, use_revin=True, patch_size=16, in_channels=7, d_model=128, num_patches=6, pred_window=48):
        super().__init__()
        self.use_revin = use_revin
        if self.use_revin:
            self.revin = RevIN(in_channels)

        self.patch_embed = PatchEmbedding(patch_size, in_channels, d_model)
        self.attention = LinearAttention(d_model)
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)

        self.feed_forward = nn.Sequential(
            nn.Linear(d_model, d_model * 2),
            nn.GELU(),
            nn.Linear(d_model * 2, d_model)
        )

        self.head = nn.Sequential(
            nn.Flatten(),
            nn.Linear(num_patches * d_model, 256),
            nn.GELU(),
            nn.Linear(256, pred_window)
        )

    def forward(self, x):
        # 1. RevIN normalization (conditionally applied)
        if self.use_revin:
            x_norm = self.revin(x, mode='norm')
        else:
            x_norm = x

        # 2. Patch embedding & linear attention processing
        patches = self.patch_embed(x_norm)
        attn_out = self.attention(patches)
        x_feat = self.norm1(patches + attn_out)
        ff_out = self.feed_forward(x_feat)
        x_feat = self.norm2(x_feat + ff_out)

        # 3. Predict output horizon
        pred_out = self.head(x_feat)

        # 4. RevIN denormalization (conditionally applied)
        if self.use_revin:
            output = self.revin(pred_out, mode='denorm')
        else:
            output = pred_out

        return output
```

```python
from torch.utils.data import TensorDataset, DataLoader
import torch.nn as nn
import torch
import numpy as np
import random

# 1. Enforce Deterministic Seeding to Prevent Session Restart Variance
def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

seed_everything(42)
print("Step 6 (Robust & Seed-Locked): Setting up DataLoaders and training...")

# 2. Convert numpy windows into PyTorch Tensors
train_dataset = TensorDataset(torch.tensor(X_train, dtype=torch.float32),
                              torch.tensor(Y_train, dtype=torch.float32))
test_dataset = TensorDataset(torch.tensor(X_test, dtype=torch.float32),
                             torch.tensor(Y_test, dtype=torch.float32))

train_loader = DataLoader(train_dataset, batch_size=128, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=128, shuffle=False)

# 3. Instantiate model, optimizer, scheduler, and move to T4 GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

model = PhysicsInformedPatchTransformerRevIN().to(device)
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)

EPOCHS = 30
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS, eta_min=1e-5)

# 4. Training Loop with Cosine Annealing & Physics Loss
model.train()
print("\nStarting Deterministic GPU Training (30 Epochs)...")

for epoch in range(EPOCHS):
    total_loss = 0.0
    for batch_x, batch_y in train_loader:
        batch_x, batch_y = batch_x.to(device), batch_y.to(device)

        optimizer.zero_grad()
        predictions = model(batch_x)
        loss = physics_informed_loss(predictions, batch_y, batch_x)

        loss.backward()
        optimizer.step()
        total_loss += loss.item() * batch_x.size(0)

    scheduler.step()
    epoch_loss = total_loss / len(train_loader.dataset)
    current_lr = scheduler.get_last_lr()[0]
    print(f"Epoch [{epoch+1}/{EPOCHS}] | Loss: {epoch_loss:.4f} | LR: {current_lr:.6f}")

# 5. Zero-Shot Cross-Station Evaluation on Dataset B (ETTm2)
model.eval()
test_loss = 0.0
criterion_mse = nn.MSELoss()

print("\nEvaluating Seed-Locked Zero-Shot Generalization on Dataset B (ETTm2)...")
with torch.no_grad():
    for batch_x, batch_y in test_loader:
        batch_x, batch_y = batch_x.to(device), batch_y.to(device)
        predictions = model(batch_x)
        mse = criterion_mse(predictions, batch_y)
        test_loss += mse.item() * batch_x.size(0)

final_test_mse = test_loss / len(test_loader.dataset)
print(f"--- Evaluation Complete ---")
print(f"Seed-Locked Zero-Shot Cross-Station Test MSE (ETTm2): {final_test_mse:.4f}")
```

```python
import torch
import torch.nn as nn
import numpy as np
import random
import pandas as pd

# Enforce strict reproducibility
def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Define ablation configurations to test
configurations = {
    "Baseline (No RevIN, No Physics)": {"revin": False, "physics": False},
    "Variant A (+ RevIN Only)": {"revin": True, "physics": False},
    "Variant B (+ Physics Loss Only)": {"revin": False, "physics": True},
    "Full Model (RevIN + Physics)": {"revin": True, "physics": True},
}

results = []
EPOCHS = 30
criterion_mse = nn.MSELoss()

print("Starting Rigorous Ablation Matrix Evaluation (Re-seeded per variant)...")

for config_name, flags in configurations.items():
    # Enforce identical initial conditions for every variant
    seed_everything(42)

    print(f"\n--- Training Configuration: {config_name} ---")

    model = PhysicsInformedPatchTransformerRevIN(
        use_revin=flags["revin"]
    ).to(device)

    optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-4)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS, eta_min=1e-5)

    # Training Loop
    model.train()
    for epoch in range(EPOCHS):
        for batch_x, batch_y in train_loader:
            batch_x, batch_y = batch_x.to(device), batch_y.to(device)
            optimizer.zero_grad()

            predictions = model(batch_x)

            if flags["physics"]:
                loss = physics_informed_loss(predictions, batch_y, batch_x)
            else:
                loss = criterion_mse(predictions, batch_y)

            loss.backward()
            optimizer.step()
        scheduler.step()

    # Zero-Shot Evaluation on ETTm2
    model.eval()
    test_loss = 0.0
    with torch.no_grad():
        for batch_x, batch_y in test_loader:
            batch_x, batch_y = batch_x.to(device), batch_y.to(device)
            predictions = model(batch_x)
            mse = criterion_mse(predictions, batch_y)
            test_loss += mse.item() * batch_x.size(0)

    final_mse = test_loss / len(test_loader.dataset)
    results.append({"Configuration": config_name, "Zero-Shot Test MSE (ETTm2)": final_mse})
    print(f"Result -> {config_name} | Test MSE: {final_mse:.4f}")

# Compile and display final comparative table for manuscript
df_results = pd.DataFrame(results)
print("\n==========================================")
print("FINAL ABLATION STUDY RESULTS FOR MANUSCRIPT")
print("==========================================")
print(df_results.to_markdown(index=False))
```
